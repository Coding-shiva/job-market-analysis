# Job Market Analytics Dashboard - Exploratory Data Analysis (EDA)

## Objective
This notebook explores the dataset of 50,000+ developer job postings across software engineering, AI/ML, data science, and full-stack development. We will discover:
1. Top in-demand technologies/skills
2. High-paying regions/cities
3. Salary distributions and factors affecting compensation (e.g. experience)
4. Relationship between remote working and listing attributes

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Setup plotting aesthetics
plt.style.use('ggplot')
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)

# Check cleaned data path
cleaned_data_path = '../data/processed/job_postings_cleaned.csv'
if not os.path.exists(cleaned_data_path):
    cleaned_data_path = 'data/processed/job_postings_cleaned.csv'
```

### 1. Load and Inspect Cleaned Data

```python
df = pd.read_csv(cleaned_data_path)
print(f"Dataset contains {df.shape[0]:,} rows and {df.shape[1]} columns.")
df.head()
```

```python
df.info()
```

### 2. Salary Analysis

```python
print("Salary Statistical Summary (USD):")
print(df['salary_avg'].describe())

# Distribution plot
plt.figure(figsize=(12, 6))
sns.histplot(df['salary_avg'], bins=40, kde=True, color='teal')
plt.axvline(df['salary_avg'].mean(), color='red', linestyle='--', label=f"Mean: ${df['salary_avg'].mean():,.2f}")
plt.axvline(df['salary_avg'].median(), color='blue', linestyle='-.', label=f"Median: ${df['salary_avg'].median():,.2f}")
plt.title('Annual Standardized Salary (USD) Distribution', fontsize=15)
plt.xlabel('Average Salary (USD)')
plt.ylabel('Posting Frequency')
plt.legend()
plt.show()
```

#### Salary by Technical Categories

```python
plt.figure(figsize=(14, 7))
sns.boxplot(x='job_category', y='salary_avg', data=df, palette='viridis')
plt.title('Salary Range by Technical Categories', fontsize=16)
plt.xlabel('Discipline')
plt.ylabel('Annual Compensation (USD)')
plt.show()
```

### 3. In-Demand Skills & Technologies

```python
# Flatten list of skills
all_skills = []
for skill_str in df['skills_cleaned'].dropna():
    all_skills.extend([s.strip() for s in skill_str.split(',') if s.strip() != ''])
    
skills_series = pd.Series(all_skills)
top_20_skills = skills_series.value_counts().head(20)

plt.figure(figsize=(14, 7))
sns.barplot(x=top_20_skills.values, y=top_20_skills.index, palette='crest')
plt.title('Top 20 Most Demanded Skills/Technologies in Job Postings', fontsize=16)
plt.xlabel('Volume of Job Openings')
plt.ylabel('Skill / Technology')
plt.show()
```

### 4. Experience vs Salary Analysis

```python
plt.figure(figsize=(12, 7))
sns.lineplot(x='experience_years', y='salary_avg', hue='job_category', data=df, ci=None, marker='o')
plt.title('Salary Scale vs Required Experience by Job Category', fontsize=16)
plt.xlabel('Years of Experience Required')
plt.ylabel('Average Compensation (USD)')
plt.legend(title='Category')
plt.show()
```

### 5. Work Setting and Regional Insights

```python
# Remote Ratio
remote_counts = df['job_type'].value_counts()

plt.figure(figsize=(8, 8))
plt.pie(remote_counts.values, labels=remote_counts.index, autopct='%1.1f%%', colors=['#4facfe', '#a78bfa', '#f35588'], startangle=140)
plt.title('Work Arrangement Distributions (Remote vs Hybrid vs Onsite)', fontsize=15)
plt.show()
```

```python
# Top 10 Hiring Cities
top_cities = df['location_city'].value_counts().head(10)

plt.figure(figsize=(12, 6))
sns.barplot(x=top_cities.values, y=top_cities.index, palette='coolwarm')
plt.title('Top 10 Hiring Tech Hubs', fontsize=15)
plt.xlabel('Total Open Postings')
plt.ylabel('Hiring Center')
plt.show()
```

### 6. Conclusions & Key Takeaways
1. **AI/ML Positions command the highest entry-level and staff salary premiums**, driven by heavy requirements in deep learning, LLMs, and PyTorch.
2. **Experience scales linearly with salary**, averaging a ~$10,000 USD premium per year of experience.
3. **Remote work has stabilised**, accounting for a high share in tech hubs while hybrid is preferred elsewhere.
